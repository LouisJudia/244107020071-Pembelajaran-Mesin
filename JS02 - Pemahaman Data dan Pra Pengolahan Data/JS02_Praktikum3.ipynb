Langkah 1 - Load Data

```python
import pandas as pd
dpath ='/content/sample_data/Titanic-Dataset-fixed.csv'
df = pd.read_csv(dpath)
df.head()
```

Langkah 2 - Slice Data

```python
df = df[['Survived', 'Pclass', 'Age', 'Sex', 'Cabin']]
df.head()
```

Langkah 3 - Encoding

```python
from sklearn.preprocessing import LabelEncoder

# melakukan encoding pada data-data kategorikal, yaitu "Sex" dan "Cabin".

le = LabelEncoder() # membuat objek dari LabelEncoder
df['Sex'] = le.fit_transform(df['Sex']) # proses encoding
df['Cabin'] = le.fit_transform(df['Cabin']) # proses encoding
```

Langkah 4 - Verifikasi Hasil

```python
# Cek kembali dengan df.head().
df.head()
```

Langkah 5 - Standarisasi

```python
from sklearn.preprocessing import StandardScaler

# melakukan standarisasi pada "Age".

std = StandardScaler()
df['Age'] = std.fit_transform(df[['Age']])
```

Kegunaan Masing-Masing Library

a. PyPREP

PyPREP adalah implementasi Python dari Preprocessing Pipeline (PREP) yang dirancang khusus untuk memproses data EEG (Electroencephalogram). Library ini menyediakan alur kerja standar untuk pembersihan awal sinyal EEG, termasuk deteksi dan interpolasi kanal yang buruk, penghilangan noise, dan referensi ulang. Tujuannya adalah memastikan konsistensi dan kualitas data EEG sebelum analisis lebih lanjut, sehingga peneliti dapat fokus pada interpretasi data tanpa terbebani oleh proses pra-pemrosesan yang rumit.

b. Scipy

SciPy (dibaca "Sigh Pie") adalah library open-source fundamental untuk komputasi ilmiah dan teknis di Python. Library ini dibangun di atas NumPy dan menyediakan berbagai modul untuk tugas-tugas seperti statistik, optimasi, integrasi, aljabar linear, transformasi Fourier, pemrosesan sinyal dan gambar, serta pemecah persamaan diferensial. SciPy sangat penting untuk analisis data ilmiah, termasuk dalam pengolahan sinyal biomedis seperti EEG dan ECG.

c. wandb (Weights & Biases)

wandb adalah platform untuk pelacakan eksperimen machine learning. Library ini memungkinkan Anda untuk mencatat metrik (seperti loss dan akurasi) secara otomatis selama pelatihan model, memvisualisasikannya dalam dashboard real-time, membandingkan beberapa run dengan hyperparameter berbeda, serta mengelola versi dataset dan model. Dengan wandb, eksperimen menjadi lebih reproducible dan kolaboratif.

d. pyECG

pyECG adalah toolkit Python open-source untuk membaca dan menulis format dataset ECG yang umum, seperti Physiobank WFDB (.hea/.dat) dan file Holter iShine (.ecg/.ann). Library ini juga menyediakan fungsionalitas untuk pemrosesan sinyal ECG, termasuk deteksi detak jantung, klasifikasi detak jantung, dan klasifikasi aritmia, sehingga memudahkan peneliti dalam membangun model machine learning untuk analisis ECG

Deskripsi Tugas
Pada tugas pratikum ini Anda akan menggunakan data "Wisconsin Breast Cancer". Data tersebut terdiri dari 569 data yang digunakan untuk mendiagnonis jenis kanker Malignant (M) dan Benign (B).


 Load Data

```python
import pandas as pd
dpath = '/content/sample_data/wbc.csv'
df = pd.read_csv(dpath)
df.head()
```

### **1. Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.**


```python
print('Informasi awal DataFrame:')
display(df.info())
```

Dari output `df.info()` terlihat bahwa kolom `Unnamed: 32` memiliki 0 entri non-null, yang berarti seluruh kolom berisi nilai `NaN`. Kolom `id` adalah identifikasi unik yang tidak relevan untuk analisis model. Oleh karena itu, kedua kolom ini akan dihapus.

```python
# Memisahkan variabel yang dapat digunakan dan yang tidak dapat digunakan

# Kolom yang tidak digunakan: 'id' dan 'Unnamed: 32'
unusable_columns = ['id', 'Unnamed: 32']

# Membuat DataFrame baru dengan kolom yang dapat digunakan (fitur dan target)
df_cleaned = df.drop(columns=unusable_columns)

print('DataFrame setelah menghapus kolom yang tidak digunakan:')
display(df_cleaned.head())
```

Sekarang, kita akan memisahkan variabel target (`diagnosis`) dari variabel fitur.

```python
# Memisahkan variabel target (y) dan variabel fitur (X)
X = df_cleaned.drop(columns=['diagnosis'])
y = df_cleaned['diagnosis']

print('\nVariabel Fitur (X) - 5 baris pertama:')
display(X.head())

print('\nVariabel Target (y) - 5 baris pertama:')
display(y.head())
```

### **2. Encoding pada Kolom 'diagnosis'**

```python
from sklearn.preprocessing import LabelEncoder

# Inisialisasi LabelEncoder
le = LabelEncoder()

# Lakukan encoding pada kolom 'diagnosis' (variabel y)
y_encoded = le.fit_transform(y)

print('Variabel Target (y) setelah encoding - 5 baris pertama:')
display(y_encoded[:5])
print('\nLabel asli dan nilai terenkode:')
for original_label, encoded_value in zip(le.classes_, le.transform(le.classes_)):
    print(f'{original_label}: {encoded_value}')
```

### **3. Standardisasi Kolom Numerik**

```python
from sklearn.preprocessing import StandardScaler

# Memilih kolom numerik untuk standardisasi
# Variabel X sudah berisi kolom-kolom numerik, kecuali 'diagnosis' yang sudah dihapus.
numeric_cols = X.select_dtypes(include=['float64', 'int64']).columns

# Inisialisasi StandardScaler
scaler = StandardScaler()

# Lakukan standardisasi pada kolom numerik di X
X_scaled = X.copy() # Buat salinan agar tidak mengubah X asli
X_scaled[numeric_cols] = scaler.fit_transform(X_scaled[numeric_cols])

print('Variabel Fitur (X) setelah standardisasi - 5 baris pertama:')
display(X_scaled.head())

print('\nStatistik deskriptif setelah standardisasi:')
display(X_scaled.describe())
```
